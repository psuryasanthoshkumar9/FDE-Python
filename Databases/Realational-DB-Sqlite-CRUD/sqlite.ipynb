```python
import sqlite3
# sqlite3 is built-in python module 
#it provide functionalities to cimmunicate with SQlite Database
```

```python
## DB connection ##
conn= sqlite3.connect("dev.db")
print("Database conntected succesfully")
#if dev.db not exists it will create automatically 
#connect() is a function 
#conn is variable  it store conttection obejects
```

```python
cur = conn.cursor()
#cursor() is instance method and cerates object
#cur is variable 
# curosor purpose is to excute SQL quries
```

```python
##Creation of Table ##
cur.execute ("""
    CREATE TABLE students (
        id INTEGER PRIMARY  KEY,
        name Text,
        age INTEGER,
        course TEXT
    )
    """)
conn.commit()

# Execute SQL statement 
# commit() sav echnages permanently
```

```python
print(type(cur.execute))
```

```python
#excuting queries ##
## First Record ##
cur.execute ("""
    INSERT INTO students (name, age, course)
    VALUES ('veera', 30, 'python')
    """)
conn.commit()
```

```python
#dynamic insertion#
## Second insert ##
query = """
    INSERT INTO students (name, age, course)
    VALUES (?, ?, ?)
    """
name = "veera"
age= 30
course = "python"
    
cur.execute (query, (name, age, course) )
conn.commit()
```

```python
cur.execute("select * from students")
records = cur.fetchall()
print(records)
```
